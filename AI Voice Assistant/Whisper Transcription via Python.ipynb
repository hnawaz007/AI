```python
from kokoro import KPipeline
```

```python
import sounddevice as sd
import warnings
# Suppress specific categories
warnings.filterwarnings("ignore", category=UserWarning)
warnings.filterwarnings("ignore", category=FutureWarning)
```

```python
MODEL_NAME = "small"
repo_id='hexgrad/Kokoro-82M'
# Initialize Kokoro
pipeline = KPipeline(lang_code='a', repo_id=repo_id)
```

```python
#speak converted audio to text
def speak(text):
    generator = pipeline(text, voice='af_heart')

    # Play and save each chunk
    for i, (gs, ps, audio) in enumerate(generator):
        print(f"Chunk {i}: Graphemes: {gs}, Phonemes: {ps}")
        
        # Play audio chunk
        sd.play(audio, samplerate=24000)
        sd.wait()  # Wait until this chunk finishes playing
```

```python
import vosk
```

```python
model = vosk.Model("vosk-model-en-us-0.42-gigaspeech")
```

```python
# Initialize the recognizer with the model
recognizer = vosk.KaldiRecognizer(model, 16000)
```

```python
# Sample audio file for recognition
audio_file = "output.raw"
```

```python
from pydub import AudioSegment
sound = AudioSegment.from_mp3(audio_file)
sound.export("sample2.wav", format="wav")
```

```python
# Open the audio file
with open(audio_file, "rb") as audio:
    while True:
        # Read a chunk of the audio file
        data = audio.read(4000)
        if len(data) == 0:
            break
        # Recognize the speech in the chunk
        recognizer.AcceptWaveform(data)
# Get the final recognized result
result = recognizer.FinalResult()
print(result)
```
